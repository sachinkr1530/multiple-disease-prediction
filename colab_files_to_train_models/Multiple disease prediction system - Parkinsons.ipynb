Importing the Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
```

Data Collection & Analysis

```python
# loading the data from csv file to a Pandas DataFrame
parkinsons_data = pd.read_csv('../dataset/parkinsons.csv')
```

```python
# printing the first 5 rows of the dataframe
parkinsons_data.head()
```

```python
# number of rows and columns in the dataframe
parkinsons_data.shape
```

```python
# getting more information about the dataset
parkinsons_data.info()
```

```python
# checking for missing values in each column
parkinsons_data.isnull().sum()
```

```python
# getting some statistical measures about the data
parkinsons_data.describe()
```

1  --> Parkinson's Positive

```python
numeric_data = parkinsons_data.select_dtypes(include=['number'])

# Grouping by 'status' and calculating the mean
grouped_mean = numeric_data.groupby('status').mean()

# Displaying the result
print(grouped_mean)
```

```python

# Plotting distributions
plt.figure(figsize=(15, 10))
plotnumber = 1

for column in numeric_data.columns:
    if plotnumber <= 30:
        ax = plt.subplot(5, 6, plotnumber)
        sns.histplot(parkinsons_data[column], kde=True, ax=ax, color='green')  # Using histplot instead of distplot (deprecated)
        plt.xlabel(column)
        plotnumber += 1

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
sns.pairplot(parkinsons_data, hue='status', diag_kind='kde', corner=True)
plt.title('Pairwise Relationships in Parkinson\'s Dataset')
plt.show()
```

```python
# Compute correlation matrix
corr_matrix = numeric_data.corr()

# Plot heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, cmap='coolwarm', annot=True, fmt='.2f', linewidths=0.5)
plt.title('Feature Correlation in Parkinson\'s Dataset')
plt.show()
```

Data Pre-Processing

Separating the features & Target

```python
X = parkinsons_data.drop(columns=['name','status'], axis=1)
Y = parkinsons_data['status']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the data to training data & Test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Model Training

Support Vector Machine Model

```python
model = svm.SVC(kernel='linear')
```

```python
# training the SVM model with training data
model.fit(X_train, Y_train)
```

Model Evaluation

Accuracy Score

```python
# accuracy score on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
```

```python
print('Accuracy score of training data : ', training_data_accuracy)
```

```python
# accuracy score on training data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
```

```python
print('Accuracy score of test data : ', test_data_accuracy)
```

```python
from sklearn.metrics import confusion_matrix


def plot_confusion_matrix(y,X_test_prediction):
    

    cm = confusion_matrix(y, X_test_prediction)
    ax= plt.subplot()
    sns.heatmap(cm, annot=True, ax = ax); #annot=True to annotate cells
    ax.set_xlabel('Predicted labels')
    ax.set_ylabel('True labels')
    ax.set_title('Confusion Matrix'); 
    # ax.xaxis.set_ticklabels(['non-cancerous', 'cancer']); ax.yaxis.set_ticklabels(['non-cancerous', 'cancer']) 
    plt.show() 
```

```python
plot_confusion_matrix(Y_test,X_test_prediction)
```

Building a Predictive System

```python
input_data = (197.07600,206.89600,192.05500,0.00289,0.00001,0.00166,0.00168,0.00498,0.01098,0.09700,0.00563,0.00680,0.00802,0.01689,0.00339,26.77500,0.422229,0.741367,-7.348300,0.177551,1.743867,0.085569)

# changing input data to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the numpy array
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

prediction = model.predict(input_data_reshaped)
print(prediction)


if (prediction[0] == 0):
  print("The Person does not have Parkinsons Disease")

else:
  print("The Person has Parkinsons")
```
